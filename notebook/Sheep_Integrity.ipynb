```python
import pandas as pd
import matplotlib.pyplot as plt
import os

results_csv = '/content/runs/detect/train/results.csv'

if os.path.exists(results_csv):
    results = pd.read_csv(results_csv)
    results.columns = results.columns.str.strip()

    plt.figure(figsize=(10, 6))
    plt.plot(results['metrics/mAP50(B)'], label='mAP @ 50', color='blue', linewidth=2)
    plt.fill_between(range(len(results)), results['metrics/mAP50(B)'], color='blue', alpha=0.1)

    plt.title('Stage 1: Detection Accuracy (mAP)', fontsize=15)
    plt.xlabel('Epochs', fontsize=12)
    plt.ylabel('Precision Score', fontsize=12)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.savefig('stage1_accuracy_graph.png', dpi=300)
    plt.show()
else:
    print("CSV not found. Ensure the path to /train/results.csv is correct!")
```
